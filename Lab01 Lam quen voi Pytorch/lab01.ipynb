### Bài 9 

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 20
np.random.seed(42)
X = np.random.rand(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([X * X, X, np.ones_like(X)])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

for epoch in range(2000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients w:\n", w.detach().numpy().flatten())
```

### Bài 10

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for epoch in range(5000):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

### Bài 11

```python
import torch
import numpy as np

print("=== YÊU CẦU 1 ===")
X = torch.arange(24)
print("Reshape 4x6:\n", X.reshape(4, 6))
print("\nReshape 2x12:\n", X.reshape(2, 12))
print("\nReshape 2x3x4:\n", X.reshape(2, 3, 4))

print("\n=== YÊU CẦU 2 ===")
A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("A + B:\n", A + B)
print("A * B:\n", A * B)
print("A / B:\n", A / B)
print("Trung bình theo cột A:\n", torch.mean(A, dim=0))
print("Độ lệch chuẩn theo cột A:\n", torch.std(A, dim=0))

print("\n=== YÊU CẦU 3 ===")
# y = 2*x^2 + 5*x + 3 tại x = 2
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()
print("Gradient dy/dx bằng PyTorch tại x=2:", x.grad.item())

# Tính thủ công: dy/dx = 4x + 5 -> tại x=2: 4*2 + 5 = 13
manual_grad = 4 * 2 + 5
print("Gradient dy/dx tính thủ công tại x=2:", manual_grad)

print("\n=== YÊU CẦU 4 ===")
# Huấn luyện đa thức y = 2x^2 - 3x + 5
poly = np.poly1d([2, -3, 5])
N = 30
np.random.seed(42)
X_np = np.random.rand(N, 1) * 5
Y_np = poly(X_np)

XX = np.hstack([X_np**2, X_np, np.ones_like(X_np)])
w_app = torch.randn(3, 1, requires_grad=True)
x_tensor = torch.tensor(XX, dtype=torch.float32)
y_tensor = torch.tensor(Y_np, dtype=torch.float32)

optimizer_app = torch.optim.NAdam([w_app], lr=0.01)

for epoch in range(3000):
    optimizer_app.zero_grad()
    y_pred = x_tensor @ w_app
    loss = torch.mean(torch.square(y_tensor - y_pred))
    loss.backward()
    optimizer_app.step()

print("Hệ số tìm được cho đa thức y = 2x^2 - 3x + 5:\n", w_app.detach().numpy().flatten())
```
